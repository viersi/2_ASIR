```python
#Pide al usuario nombres y edades hasta escribir 'fin'. 
#Guarda los datos en un diccionario. Luego muestra la edad media.
fin = False
d = {

}
#Bucle principal
print("Escribe 'fin' en cualquier momento para cerrar el programa")
while fin != True:
    usrn = input("Escribe un nombre: ")
    if usrn == "fin":
        fin = True
        break
    usra = input("Escribe una edad para el anterior nombre: ")
    if usra == "fin":
        fin = True
        break
    usra = int(usra)
#Cuando se han introducido clave y valor, se actualiza el diccionario.
    d.update({usrn:usra})
j = 0
l = 0
#Bucle para hacer la media de edad
for i in d:
    j += d[i]
    l += 1
res = j//l
print(f"Edad media: {res}")
```

```python
#Dado un texto, muestra cuántas veces aparece cada letra usando un diccionario de frecuencias. 
d = {

}

texto = "Lorem ipsum dolor etc etct murcielago valencia Padreada QuEYo Seyyyyy"

for i in texto.lower():
    if i == " " or i in "0123456789":
        continue
    elif i not in d:
        d.update({i:1})
    else:
        d[i] += 1

for i, j in d.items():
    print(i, j)
```

```python
#Combina dos diccionarios `a` y `b` en uno solo llamado `c` usando `update()` o `{**a, **b}`.
a = {
    'x': 2,
    'i': 4,
    'j': 6,
    'k': 5
}
b = {
    'a': 1,
    'y': 3,
    'j': 4,
    'z': 3
}

#Con esta sintaxis se puede crear un nuevo diccionario. Si hay claves comunes, se toman los valores del segundo diccionario.
c = {**a, **b}
#.update() no crea nuevos diccionarios, asi que lo copiamos primero con el metodo .copy(). 
#Si hay claves comunes, se toman los valores del segundo diccionario.
d = a.copy()
d.update(b)

print(c)
print(d)
```

```python
#Crea una lista de diccionarios con los datos de 3 empleados 
#(`nombre`, `salario`, `departamento`). 
#Muestra el salario promedio y filtra los del departamento 'Ventas'.
listadic = [
    {'nombre': 'Victor', 'salario': 40000, 'departamento': 'IT'},
    {'nombre': 'Andrea', 'salario': 30000, 'departamento': 'Ventas'},
    {'nombre': 'Jenny', 'salario': 35000, 'departamento': 'Ventas'}
]

sal_prom = 0
deps = 0

for i in listadic:
    if i['departamento'] == 'Ventas':
        sal_prom += i['salario']
        deps += 1

sal_prom /= deps

print(sal_prom)
```

```python
#Crea un diccionario anidado que represente un aula y calcula la nota media de sus alumnos.
aula = {
    'grado': 'ASIR',
    'curso': '2º',
    'victor': {
        'nota_media': 9,
        'asistencia': True
    },
    'diego': {
        'nota_media': 7,
        'asistencia': False
    },
    'jenny': {
        'nota_media': 9,
        'asistencia': True
    },
    'andrea': {
        'nota_media': 8,
        'asistencia': True
    }
}

avg = 0
c = 0

for i, j in aula.items():
    if isinstance(j, dict):
        avg += j['nota_media']
        c += 1

avg /= c
print(avg)
```

```python
#Usa un diccionario para contar cuántas veces aparece cada palabra en una 
#frase introducida por el usuario.
d = {

}

usr = input()
#El metodo .split() separa una cadena por palabras automaticamente, haciendo más facil hacer el recuento de palabras.
#Podemos encadenar metodos, en este caso pasamos a minusculas, y separamos por palabras.
for i in usr.lower().split():
    if i not in d:
        d.update({i:1})
    else:
        d[i] += 1
for i, j in d.items():
    print(i, j)
```

```python
#Simula una base de datos de usuarios: 
#permite registrar e iniciar sesión comprobando email y contraseña.

#Creamos el diccionario donde van a ir los pares usuario/contraseña.
usuarios = {}

#Usamos funciones para hacer más util el menu, pudiendo regresar en caso de error.
#Función login.
def ft_login():
    print("Escribe 'quit' para volver al menu")
    email = input("Introduce tu email: ").lower()
    if email == "quit":
#Usamos return para salir de cada función del programa y volver al menu.
        return
    while "@" not in email:
        email = input("Email no valido: ").lower()
        if email == "quit":
            return
    pwd = input("Introduce tu contraseña: ")
    if email in usuarios and usuarios[email] == pwd:
        print("Inicio de sesión correcto")
    else:
        print("Email o contraseña incorrectos, vuelve a introducir tus datos")
        ft_login()
    return
#Función de registro.    
def ft_register():
    print("Escribe 'quit' para volver al menu")
    email = input("Introduce un email: ")
    if email == "quit":
        return
    while "@" not in email:
        email = input("Email no valido: ")
        if email == "quit":
            return
    pwd = input("Introduce una contraseña: ")
    usuarios.update({email:pwd})
    print("Usuario registrado exitosamente\n")
    return
#Función principal.    
def ft_menu():
#Esta función se ejecuta indefinidamente hasta que el usuario decide salir.
    while True:
        print("(1) Login | (2) Registro | Marca cualquier otro caracter para salir.")
        print("Solo desarrolladores: (x) Mostrar base de datos \n")
        select = input("¿Qué deseas hacer?: \n")
        if select == "1":
            print("##LOGIN##")
            ft_login()
        elif select == "2":
            print("##REGISTRO##")
            ft_register()
#Esta opción se incluye solamente para comprobar que el registro está funcionando correctamente.
        elif select == "x":
            print("##USUARIOS REGISTRADOS##")
            print(usuarios)
        else:
            print("Hasta la próxima")
#En este caso usamos return para detener el bucle y salir de la función.
            return

ft_menu()
```
