```python
#Programa de ventas que permite añadir productos, tramitar pedidos, y generar informes de ventas.

#Generamos listas con diccionarios para nuestros productos y los pedidos. Productos empieza con diccionarios dentro,
#pedidos no, ya que se ira llenando con los pedidos del usuario.
products = [
    {"id": "0","nombre": "NvME","precio": 100,"stock": 50},
    {"id": "1","nombre": "RAM","precio": 120,"stock": 67}
]

orders = []

#Generamos una funcion independiente para cada funcionalidad del programa. Las funciones interactivas
#se pueden detener con 'quit' para volver al menu.

#Función para mostrar el catalogo
def ft_catalog():
    print("\n\nID  NOMBRE  PRECIO  STOCK\n\n")
    for i in products:
        print(i["id"], i["nombre"], i["precio"], i["stock"])
    print("\n")
#Función para mostrar el historial de pedidos
def ft_order_list():
    print("\n\nPRODUCTO  CANTIDAD  PRECIO  PRECIO TOTAL\n\n")
    for i in orders:
        print(i["producto"], i["cantidad"], i["precio_unitario"], i["total"])
    print("\n")
#Función para añadir productos al catalogo.        
def ft_addproduct():
    print("Escribe 'quit' para volver al menu")
#Comprobamos que el ID no exista en el catalogo.
    pid = input("Introduce el nuevo ID: ")
    if pid == "quit":
        return
#Se comprueba con el metodo .isdigit() que el input sea exclusivamente numerico.
    while pid.isdigit() == False:
        pid = input("El ID debe ser numerico: ")
        if pid == "quit":
            return
#Con la funcion any() y un bucle for comprobamos que el ID introducido no coincida con ningun valor del
#indice 'id' en nuestra lista de diccionarios.
    while any(i['id'] == pid for i in products):
        pid = input("ID no disponible: ")
        if pid == "quit":
            return
#Registramos el nombre.        
    pname = input("Introduce el nombre del producto: ")
    if pname == "quit":
            return
#Comprobamos que el precio sea númerico y mayor que cero.    
    pprice = input("Introduce el preio unitario: ")
    while pprice.isdigit() == False:
        pprice = input("El precio debe ser un numero: ")
        if pprice == "quit":
            return
    while int(pprice) <= 0:
        pprice = input("El precio debe estar por encima de cero: ")
        if pprice == "quit":
            return
#Comprobamos que el stock sea númerico y mayor que cero.            
    pstock = input("Introduce el stock del producto: ")
    while pstock.isdigit() == False:
        pstock = input("El stock debe ser un numero: ")
        if pstock == "quit":
            return
    while int(pstock) <= 0:
        pstock = input("El stock debe estar por encima de cero: ")
        if pstock == "quit":
            return
#Guardamos los inputs en un diccionario y formateamos las variables necesarias.        
    new_dict = {
        "id": pid,
        "nombre": pname,
        "precio": int(pprice),
        "stock": int(pstock)
    }
#Con el metodo .append() añadimos el nuevo diccionario a nuestro catalogo.   
    products.append(new_dict)
    
    return
#Función para realizar pedidos.
def ft_order():
#Mostramos el catalogo para que el usuario pueda ver el id del producto a pedir.
    ft_catalog()
#Creamos esta variable vacia para guardar en ella un enlace al catalogo.
#Este enlace es muy util, ya que nos permite más adelante restar la cantidad de un producto
#accediendo directamente a su valor en el catalogo.
    sel_product = 0
    
    print("Escribe 'quit' para volver al menu")
#Comprobamos que el id exista en el catalogo.
    pid = input("Introduce el ID del producto: ")
    if pid == "quit":
        return
    while pid.isdigit() == False:
        pid = input("El ID debe ser numerico: ")
        if pid == "quit":
            return
    while not any(i['id'] == pid for i in products):
        pid = input("ID no existente: ")
        if pid == "quit":
            return
#Utilizamos el id proporcionado para encontrar el diccionario correcto en el catalogo,
#y alamacenamos su referencia en sel_product, pudiendo asi modificar el valor de sus claves
#"en tiempo real".
    for i in products:
        if pid == i["id"]:
            sel_product = i
#Comprobamos que haya suficiente stock para poder tramitar el pedido.        
    oquantity = input("Selecciona la cantidad deseada: ")
    while oquantity.isdigit() == False:
        oquantity = input("La cantidad debe ser numerica: ")
        if oquantity == "quit":
            return
    while int(oquantity) <= 0:
        oquantity = input("La cantidad debe estar por encima de cero: ")
        if oquantity == "quit":
            return
    while int(oquantity) > sel_product["stock"]:
        oquantity = input("No existe suficiente stock: ")
        if oquantity == "quit":
            return
#Creamos un nuevo diccionario con la información del pedido.
    new_dict = {
        "producto": sel_product["nombre"],
        "cantidad": int(oquantity),
        "precio_unitario": sel_product["precio"],
        "total": int(oquantity) * sel_product["precio"]
    }
#Añadimos el pedido a nuestro historial de pedidos.    
    orders.append(new_dict)
#Accedemos a traves del enlace a la clave stock del producto seleccionado,
#y le restamos la cantidad pedida por el usuario.
    sel_product['stock'] -= int(oquantity)  
    return
#Función para mostrar informe de ventas.
def ft_sales():
#Creamos variables para todo aquello que vayamos a mostrar.
#Con la funcion sum() y un bucle for podemos agrupar todos los valores de la clave
#que seleciconemos en nuestra lista de pedidos de manera mas sencilla.
    product_sales = sum(i["cantidad"] for i in orders)
    sales_total = sum(i["total"] for i in orders)
#Mostramos el total de pedidos contando los elementos de nuestra lista con la funcion len().
    total_orders = len(orders)
#Mostramos cada una de nuestras variables.    
    print(f"Productos vendidos: {product_sales}")
    print(f"Importe total generado: {sales_total}")
    print(f"Pedidos realizados: {total_orders}")

    return
    
#Función principal.    
def ft_menu():
    ft_catalog()
#Esta función se ejecuta indefinidamente hasta que el usuario decide salir.
    while True:
        print("(1) Catalogo | (2) Añadir producto | (3) Hacer pedido | (4) Ventas | (5) Historial pedidos")
        print("Selecciona cualquier otro caracter para salir \n")
        select = input("¿Qué deseas hacer?: \n")
        if select == "1":
            print("##CATALOGO##")
            ft_catalog()
        elif select == "2":
            print("##AÑADIR PRODUCTO##")
            ft_addproduct()
        elif select == "3":
            print("##HACER PEDIDO##")
            ft_order()
        elif select == "4":
            print("##VENTAS##")
            ft_sales()
        elif select == "5":
            print("##PEDIDOS##")
            ft_order_list()
        else:
            print("Hasta la próxima")
#En este caso usamos return para detener el bucle y salir de la función.
            return

ft_menu()
```
