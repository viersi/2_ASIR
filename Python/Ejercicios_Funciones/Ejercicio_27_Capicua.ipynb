```python
#Recibe un numero y devuelve si es capicua.
def ft_npalindrome(x):
#Convertimos el numero en string para poder recorrerlo.
    x = str(x)
    list = []
    list_2 = []
#Recorremos la cadena. La guardamos como lista, y como lista invertida.
    for i in x:
        list = [i] + list
    for i in x:
        list_2 = list_2 + [i]
#Si ambas listas coinciden, el numero es capicua.
    if list == list_2:
        return("Capicua")
    else:
        return("No capicua")

ft_npalindrome(8008)
```
