```python
#Recibe una frase y cuenta cuantas palabras tiene.
def ft_word_count(x):
    j = 0
#Variable para controlar si estamos dentro de una palabra.
    inword = False
#Recorremos la cadena.
    for i in x:
#Si el caracter no es un espacio, y no estabamos dentro de una palabra, aumentamos en uno el contador.
        if i != " " and inword == False:
            j += 1
#Cambiamos el estado de inword a True para que no siga sumando el contador mientras estemos dentro de la misma palabra.
            inword = True
#Al encontrar un espacio cambiamos el estado de inword a False, para que la siguiente palabra incremente el contador.
        if i == " ":
            inword = False
    return(j)
ft_word_count("frase de prueba ola")
```
