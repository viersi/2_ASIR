```python
#Recibe una contraseña y devuelve si es valida (minimo 8 caracteres, al amenos una mayuscula y un numero).
def ft_pwdpol(x):
#Se crean variables de control para cada requisito.
    length = False
    num = False
    caps = False
#Se comprueba la longitud.
    if len(x) >= 8:
        length = True
#Se comprueba el caracter numerico.
    for i in x:
        if i in "0123456789":
            num = True
#Se comprueba la mayuscula.
    for i in x:
        if i in "QWERTYUIOPASDFGHJKLÑZXCVBNM":
            caps = True
#Si todas las variables de control son True, la contraseña es valida. Si al menos una es False, la contraseña no es valida.
    if length and num and caps:
        return("Valida")
    return("No valida")

ft_pwdpol("P4ssw0rd")
```
