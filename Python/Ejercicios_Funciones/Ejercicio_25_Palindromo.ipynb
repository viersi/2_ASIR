```python
#Recibe una cadena y determina si es palindroma.
def ft_palindrome(x):
#Formateamos la cadena a minusculas para evitar problemas con los caracteres.
    x = x.lower()
    list = []
    list_2 = []
#Recorremos la cadena dos veces. La convertimos en lista, y en la lista invertida.
    for i in x:
        list = [i] + list
    for i in x:
        list_2 = list_2 + [i]
#Si ambas listas son iguales, la cadena era palindroma.
    if list == list_2:
        return("Palindromo")
    else:
        return("No palindromo")

ft_palindrome("Sos")
```
