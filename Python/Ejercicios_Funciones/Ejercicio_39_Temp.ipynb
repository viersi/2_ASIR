```python
#Recibe una lista de temperaturas y devuelve cuantas estan por encima de la media.
def ft_above_temp(x):
    list = []
#Calcula la media de temperaturas.
    avg = sum(x) / len(x)
#Recorre la lista y si el elemento es mayor que la media lo añade a la nueva lista.
    for i in x:
        if i > avg:
            list = list + [i]
    return(list)

ft_above_temp([23, 17, 29, 31, 21, 19, 26, 28, 22, 25])
```
