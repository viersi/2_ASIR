```python
#Recibe una cantidad de segundos y la devuelve formateada a horas minutos y segundos.
def ft_time(x):
#Calculamos las horas y los minutos, y guardamos los segundos que restan.
    h = x // 3600
    m = (x % 3600) // 60
    s = (x % 3600) % 60
#Almacenamos una variable con los valores obtenidos convertidos a cadenas y le damos fromato para ser mas legibles.
    time = str(h) + "h:" + str(m) + "m:" + str(s) + "s"
    return(time)

ft_time(15875)
```
