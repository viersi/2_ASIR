```python
#Calcula cuantos numeros positivos y negativos hay en una lista.
def ft_neg_pos(x):
    p = 0
    n = 0
#Se recorre la lista, si el elemento es menor que cero se incrementa en uno el contador de negativos, y si es mayor, el de los positivos.
#Cero no incrementa ningun contador, ya que no se considera ni positivo ni negativo.
    for i in x:
        if i < 0:
            n += 1
        elif i > 0:
            p += 1
#Devolvemos primero los negativos y luego los positivos.
    return(n, p)

ft_neg_pos([4, -2, 7, -1, 0, -8, 3, -5, 9, -3])
```
