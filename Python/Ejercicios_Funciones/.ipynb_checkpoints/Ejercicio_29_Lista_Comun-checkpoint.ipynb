```python
#Recibe dos listas y devuelve una lista con sus elementos comunes.
def ft_commonval(x, y):
    list = []
#Recorre la primera lista. Si el elemento se encuentra entre los elementos de la otra lista, lo añade a la lista vacia creada anteriormente.
    for i in x:
        if i in y:
            list = list + [i]

    return(list)

ft_commonval([3, 7, 2, 9, 5, 8, 10, 12, 14], [4, 9, 10, 11, 12, 15, 18])
```
