```python
#Funcion que genera un menu interactivo con opciones y funciones asociadas.
#Primero definimos todas las funciones que vamos a usar.
def ft_sum(x, y):
    return(x + y)

def ft_res(x, y):
    return(x - y)

def ft_revstr(x):
    list = []
    for i in x:
        list = [i] + list
    return("".join(list))

def ft_pyramid(x):
    c = 1
    for i in range(1, x + 1):
        print(" " * (x - i), "*" * c)
        c += 2
    return
#Funcion principal.
def ft_program():
#Declaramos esta variable para poder mostrar el return de las funciones.
    show = 0
#Mostramos al usuario las posibles opciones.
    print("(1) Suma | (2) Resta | (3) Word reverse | (4) Piramide | (X) Salir\n\n")
#Pedimos al usuario que seleccione una opción
    sel = input("Selecciona una opción: ")
#Comprobamos que opción ha escogido el usuario. Si no coincide con ninguna, se le pide otra vez.
    if sel == "1":
        print("Dame dos números para sumar n1 + n2")
        x = int(input("n1: "))
        y = int(input("n2: "))
#Almacenamos el valor que devuelve la función para poder mostrarlo.
        show = ft_sum(x, y)
        print(f"{x} + {y} = {show}")
    elif sel == "2":
        print("Dame dos números para restar n1 - n2")
        x = int(input("n1: "))
        y = int(input("n2: "))
        show = ft_res(x, y)
        print(f"{x} - {y} = {show}")
    elif sel == "3":
        print("Dame cualquier palabra o frase y le doy la vuelta")
        x = input("Tu texto: ")
        show = ft_revstr(x)
        print(show)
    elif sel == "4":
        print("Voy a generar una piramide con tantos pisos como quieras")
        x = int(input("Número de pisos: "))
        ft_pyramid(x)
#Sale de la función devolviendo null.
    elif sel == "x" or sel == "X":
        return
#Llama a la función hasta que el usuario seleccione una opción válida.
    else:
        print("Selecciona una opción válida")
        ft_program()

ft_program()
```
