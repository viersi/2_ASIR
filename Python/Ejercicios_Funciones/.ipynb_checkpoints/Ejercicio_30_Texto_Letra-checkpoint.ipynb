```python
#Recibe un texto y una letra y devuelve cuantas veces aparece la letra en el texto.
def ft_find(x, y):
    counter = 0
#Recorre el texto y si el caracter es igual al segundo parametro aumenta en uno el contador.
    for i in x:
        if i == y:
            counter += 1

    return(counter)

ft_find("En el principio, los dioses forjaron los cielos y la tierra, y de la espuma del mar nació la primera criatura inmortal.", "a")
```
