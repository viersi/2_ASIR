```python
#Recibe un numero y devuleve True si es primo, False si no.
def ft_is_prime(x):
#Recorre todos los numeros entre 2 y el anterior al numero. Si el resto entre alguno es cero, no es primo.
    for i in range(2, x):
        if x % i == 0:
            return(False)
    return(True)

ft_is_prime(5)
```
