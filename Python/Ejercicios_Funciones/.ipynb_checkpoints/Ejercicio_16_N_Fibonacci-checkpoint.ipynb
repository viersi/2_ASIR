```python
#Recibe un numero n y devuelve los n primeros terminos de la sucesion de Fibonacci.
def ft_fibo(x):
    list = []
    y = 0
    z = 1
#Itera n veces sumando el indice actual al anterior y añadiendo el resultado en la lista que sera devuelta.
    for i in range(x):
        res = y + z
        list += [res]
        y = z
        z = res
    return(list)

ft_fibo(10)
```
