```python
#Recibe un numero n y devuelve una lista con los n primeros pares.
def ft_first_evens(x):
    list = []
#Recorremos los numeros de 1 hasta el doble del numero introducido, ya que solo la mitad de los que hay hasta el numero son pares.
#De esta manera obtenemos exactamente tantos pares como el parametro introducido.
    for i in range(1, (x + 1) * 2):
        if i % 2 == 0:
            list += [i]

    return(list)

ft_first_evens(10)
```

```python
#Segunda implementacion. El doble de cualquier numero es par. Podemos multiplicar todos los numeros entre 1 y n por 2 para obtener n pares.
def ft_nevens(x):
    list = []
    for i in range(1, x + 1):
        list += [(i * 2)]
    return(list)
ft_nevens(10)
```
