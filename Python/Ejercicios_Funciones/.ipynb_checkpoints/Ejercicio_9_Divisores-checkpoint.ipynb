```python
#Recibe un numero y devuelve sus divisores.
def ft_divisor(x):
    list = []
#Recorre de 1 al numero, y si el resto de la division es cero, se añade a la lista de divisores.
    for i in range(1, x + 1):
        if x % i == 0:
            list = list + [i]
    return(list)

ft_divisor(50)
```
