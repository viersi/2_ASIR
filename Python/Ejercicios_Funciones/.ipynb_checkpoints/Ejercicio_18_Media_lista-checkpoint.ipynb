```python
#Recibe una lista de notas y calcula la media.
def ft_list_avg(x):
    j = 0
#Recorre la lista, suma todos sus elementos y guarda el valor en j, y devuelve j entre el numero de elementos de la lista.
    for i in x:
        j += i
    return(j / len(x))

ft_list_avg([7.5, 8.0, 6.5, 9.2, 5.8, 7.0, 8.7, 4.5, 6.8, 9.5])
```
