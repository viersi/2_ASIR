```python
#Recibe una lista y un valor y devuelve cuantas veces aparece ese valor en la lista.
def ft_lvalue(x, y):
    counter = 0
#Recorre la lista. Cada vez que el indice es igual al segundo parametro aumenta en uno el contador.
    for i in x:
        if i == y:
            counter += 1
    return(counter)

ft_lvalue([3, 7, 2, 3, 8, 7, 2, 9, 3, 7, 5, 2, 8, 9, 3], 7)
```
