```python
#Recibe una cadena y devuelve cuantos caracteres numericos hay en ella.
def ft_nstr(x):
    counter = 0
#Recorre la cadena y si el elemento es un numero aumenta en uno el contador.
    for i in x:
        if i in "0123456789":
            counter += 1
    return(counter)

ft_nstr("La vida es una y hay que viv1rla d4nza kudur0")
```
