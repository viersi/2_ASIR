```python
#Recibe una lista y devuelve sus valores minimo y maximo.
def ft_min_max(x):
    l = len(x)
#Recorremos la lista y usamos el algoritmo burbuja para ordenarla.
    for i in range(l):
        for j in range(0, l - i - 1):
            if x[j] > x[j + 1]:
                x[j], x[j + 1] = x[j + 1], x[j]
#Devolvemos el indice 0 y el indice anterior a la longitud de la lista.
    return(x[0], x[l - 1])

ft_min_max([5, 6, 8, 1, 2, 45, 8, 99, 25, 7])
```
