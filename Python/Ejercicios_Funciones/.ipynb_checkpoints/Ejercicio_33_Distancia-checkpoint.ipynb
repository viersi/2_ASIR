```python
#Recibe dos puntos ((x1,y1) y (x2,y2)) y devuelve la distancia entre ellos.
#Importamos la libreria math para poder calcular raices cuadradas.
import math
#La distancia entre dos vectores se obtiene aplicando el teorema de pitagoras.
def ft_distance(x, y):
#Definimos la variable d y almacenamos en ella el resultado de la formula para calcular la distancia entre dos puntos.
    d = math.sqrt((y[0] - x[0])**2 + (y[1] - x[1])**2)
#Devolvemos el resultado redondeado a dos decimales.
    return(round(d, 2))
              
ft_distance([1, 1], [3, 8])
```
