```python
#Recibe un numero de DNI y calcula la letra que le corresponde.
def ft_letradni(x):
#Creamos una lista con las posibles letras. Cada una ocupa la posicion que le corresponde, que se calcula con el resto del
#numero de DNI entre 23.
    pool = ["T", "R", "W", "A", "G", "M", "Y", "F", "P", "D", "X", "B", "N", "J", "Z", "S", "Q", "V", "H", "L", "C", "K", "E"]
    resto = x % 23
    return(pool[resto])

ft_letradni(54037169)
```
