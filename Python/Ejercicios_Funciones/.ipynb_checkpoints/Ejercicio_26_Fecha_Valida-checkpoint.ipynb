```python
#Recibe una fecha en formato dia/mes/año y devuelve si es valida.
def ft_valid_date(d, m, y):
#Creamos una variable de control para los años bisiestos.
    bis = False
#Comprobamos si el año es bisiesto.    
    if y % 4 == 0:
        bis = True
#Comprobamos mes valido.    
    if m < 1 or m > 12:
        return(False)
#Comprobamos año valido.
    if y < 1:
        return(False)
#Comprobamos dia valido.
    if d < 1:
        return(False)
#Comprobamos dia para meses de 31 dias.
    if m in {1, 3, 5, 7, 8, 10, 12} and d > 31:
        return(False)
#Comprobamos dia para meses de 30 dias.
    if m in {4, 6, 9, 11} and d > 30:
        return(False)
#Comprobamos dia para febrero en caso de que no sea y de que sea bisiesto.
    if m == 2 and d > 28:
        return(False)
    if m == 2 and bis and d > 29:
        return(False)
    return(True)

ft_valid_date(31, 12, 1999)    
```
