```python
def ft_pwdpol(x):
    length = False
    num = False
    caps = False
    if len(x) >= 8:
        length = True
    for i in x:
        if i in "0123456789":
            num = True
    for i in x:
        if i in "QWERTYUIOPASDFGHJKLÑZXCVBNM":
            caps = True
    if length and num and caps:
        return("Valida")
    return("No valida")

ft_pwdpol("M3dac2025")
```
